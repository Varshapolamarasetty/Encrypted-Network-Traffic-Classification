```python
import pandas as pd
import numpy as np
import joblib

from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight
```

```python
df = pd.read_csv("vpn_nonvpn_merged_final.csv")
```

```python
print(df.shape)
print(df.isnull().sum())
print(df["class1"].unique())
```

```python
import pandas as pd

# Count samples per class
class_counts = df['class1'].value_counts()
print(class_counts)
```

```python
from imblearn.over_sampling import SMOTE
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from xgboost import XGBClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
```

```python
# ----------------------------
# 1. Separate features and target
# ----------------------------
X = df.drop('class1', axis=1)  # numeric features
y = df['class1']               # target labels (strings)
```

```python
# ----------------------------
# 2. Apply SMOTE to balance classes
# ----------------------------
smote = SMOTE(random_state=42)
X_res, y_res = smote.fit_resample(X, y)

print("\nBefore SMOTE:\n", class_counts)
print("\nAfter SMOTE:\n", pd.Series(y_res).value_counts())
```

```python
from sklearn.preprocessing import LabelEncoder
import joblib

# ----------------------------
# Label encode the target
# ----------------------------
le = LabelEncoder()
y_res_enc = le.fit_transform(y_res)  # numeric labels

# ----------------------------
# Print the mapping of classes → numbers
# ----------------------------
print("Class → Encoded Label Mapping:")
for cls, enc in zip(le.classes_, range(len(le.classes_))):
    print(f"{cls} → {enc}")

# ----------------------------
# Save the LabelEncoder for later
# ----------------------------
joblib.dump(le, 'label_encoder.pkl')
print("\nLabelEncoder saved as 'label_encoder.pkl'")
```

```python
# ----------------------------
# 4. Split into train/test sets
# ----------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X_res, y_res_enc, test_size=0.2, random_state=42, stratify=y_res_enc
)
```

```python
# ----------------------------
# 5. Train XGBoost Classifier
# ----------------------------
xgb_clf = XGBClassifier(use_label_encoder=False, eval_metric='mlogloss', random_state=42)
xgb_clf.fit(X_train, y_train)
y_pred_xgb = xgb_clf.predict(X_test)

print("\nXGBoost Classification Report:")
print(classification_report(y_test, y_pred_xgb, target_names=le.classes_))
```

```python
# ----------------------------
# 6. Train Random Forest Classifier
# ----------------------------


rf_clf = RandomForestClassifier(n_estimators=5, random_state=42)
rf_clf.fit(X_train, y_train)
y_pred_rf = rf_clf.predict(X_test)

print("\nRandom Forest Classification Report:")
print(classification_report(y_test, y_pred_rf, target_names=le.classes_))
```

```python
import joblib
from xgboost import XGBClassifier
from sklearn.ensemble import RandomForestClassifier

# ----------------------------
# 1. Save the trained models
# ----------------------------
joblib.dump(xgb_clf, 'xgb_vpn_model.pkl')
print("XGBoost model saved as 'xgb_vpn_model.pkl'")

joblib.dump(rf_clf, 'rf_vpn_model.pkl')
print("Random Forest model saved as 'rf_vpn_model.pkl'")

# ----------------------------
# 2. Save the LabelEncoder
# ----------------------------
joblib.dump(le, 'label_encoder.pkl')
print("LabelEncoder saved as 'label_encoder.pkl'")

# ----------------------------
# 3. Optional: Save class mapping
# ----------------------------
class_mapping = dict(zip(le.classes_, range(len(le.classes_))))
joblib.dump(class_mapping, 'class_mapping.pkl')
print("Class mapping saved as 'class_mapping.pkl'")
print("Class → Label mapping:", class_mapping)

```

```python
import gradio as gr
import pandas as pd
import joblib
import numpy as np

# ----------------------------
# 1. Load saved model, encoder, and dataset
# ----------------------------
xgb_model = joblib.load('xgb_vpn_model.pkl')   # XGBoost model
le = joblib.load('label_encoder.pkl')          # LabelEncoder
df = pd.read_csv("/content/vpn_nonvpn_merged_final.csv")            # Your dataset with all features + 'class1'

feature_names = [col for col in df.columns if col != 'class1']

# ----------------------------
# 2. Prediction function
# ----------------------------
def predict_random_sample():
    # Pick a random row from the dataset
    sample = df.sample(1, random_state=np.random.randint(0, 1000))

    X_input = sample[feature_names]

    # Predict encoded label
    y_pred_enc = xgb_model.predict(X_input)

    # Convert numeric label back to class name
    y_pred_label = le.inverse_transform(y_pred_enc)

    # Return features, true class, and predicted class
    return X_input.values[0], sample['class1'].values[0], y_pred_label[0]

# ----------------------------
# 3. Build Gradio Interface
# ----------------------------
iface = gr.Interface(
    fn=predict_random_sample,
    inputs=[],
    outputs=[
        gr.Dataframe(headers=feature_names, label="Random Sample Features"),
        gr.Textbox(label="True Class"),
        gr.Textbox(label="Predicted Class")
    ],
    title="VPN Traffic Classifier - Random Sample",
    description="Click submit to see a random sample from the dataset and its predicted VPN traffic class."
)

# ----------------------------
# 4. Launch the app
# ----------------------------
iface.launch()
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, LSTM, Conv1D, MaxPooling1D, Flatten, Dropout
from tensorflow.keras.utils import to_categorical
```

```python
# Convert X_res to NumPy array
X_res_np = X_res.values  # pandas DataFrame → NumPy array

# Reshape for LSTM/CNN: (samples, timesteps=1, features)
X_res_3d = X_res_np.reshape((X_res_np.shape[0], 1, X_res_np.shape[1]))

# One-hot encode target for deep learning
y_res_cat = to_categorical(y_res_enc)

# Split into train/test
X_train_dl, X_test_dl, y_train_dl, y_test_dl = train_test_split(
    X_res_3d, y_res_cat, test_size=0.2, random_state=42, stratify=y_res_enc
)
```

```python
X_res_np
```

```python
from tensorflow.keras.layers import Bidirectional

lstm_model = Sequential([
    Bidirectional(
        LSTM(128, return_sequences=True),
        input_shape=(X_train_dl.shape[1], X_train_dl.shape[2])
    ),
    Dropout(0.3),

    Bidirectional(LSTM(64)),
    Dropout(0.3),

    Dense(64, activation='relu'),
    Dropout(0.2),
    Dense(y_train_dl.shape[1], activation='softmax')
])

lstm_model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

lstm_model.summary()

history_lstm = lstm_model.fit(
    X_train_dl, y_train_dl,
    validation_data=(X_test_dl, y_test_dl),
    epochs=30,
    batch_size=32
)
```

```python
cnn_model = Sequential([
    Conv1D(64, kernel_size=1, activation='relu', input_shape=(X_train_dl.shape[1], X_train_dl.shape[2])),
    MaxPooling1D(pool_size=1),
    Dropout(0.2),
    Flatten(),
    Dense(32, activation='relu'),
    Dense(y_train_dl.shape[1], activation='softmax')
])

cnn_model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
cnn_model.summary()

# Train CNN
history_cnn = cnn_model.fit(
    X_train_dl, y_train_dl,
    validation_data=(X_test_dl, y_test_dl),
    epochs=20, batch_size=32
)
```

```python
from sklearn.metrics import classification_report
import numpy as np

# LSTM
y_pred_lstm = lstm_model.predict(X_test_dl)
y_pred_labels_lstm = np.argmax(y_pred_lstm, axis=1)
y_true_labels = np.argmax(y_test_dl, axis=1)

print("\nLSTM Classification Report:")
print(classification_report(y_true_labels, y_pred_labels_lstm, target_names=le.classes_))

# CNN
y_pred_cnn = cnn_model.predict(X_test_dl)
y_pred_labels_cnn = np.argmax(y_pred_cnn, axis=1)

print("\nCNN Classification Report:")
print(classification_report(y_true_labels, y_pred_labels_cnn, target_names=le.classes_))
```
